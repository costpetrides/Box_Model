```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import pandas as pd
import seaborn as sns

# 1. Load the CSV file from GitHub
url = 'https://raw.githubusercontent.com/costpetrides/Box_Model/refs/heads/main/TESTout1-Table%201.csv'

# Note: sep=';' for semicolon separation, decimal=',' for comma-separated decimals
df = pd.read_csv(url, sep=';', decimal=',')

# Clean up column names (removes trailing/leading spaces like in 'Time ')
df.columns = df.columns.str.strip()

# 2. Convert Time (in minutes) to Hours of the Day
# Change start_hour if your simulation starts at a different time (e.g., 6 for 06:00)
start_hour = 0
time_in_hours = start_hour + (df['Time'] / 60)

# 3. Set up the plotting style
sns.set_theme(style='whitegrid')
fig, ax1 = plt.subplots(figsize=(12, 6))

# Determine marker frequency dynamically so lines aren't overcrowded
marker_interval = max(1, len(df) // 20)

# 4. Primary y-axis: Plot O3, NO, NO2
ax1.plot(
    time_in_hours,
    df['O3'],
    label=r'O$_3$',
    color='tab:blue',
    linewidth=2,
    marker='o',
    markevery=marker_interval,
)
ax1.plot(
    time_in_hours,
    df['NO'],
    label='NO',
    color='tab:orange',
    linewidth=2,
    marker='s',
    markevery=marker_interval,
)
ax1.plot(
    time_in_hours,
    df['NO2'],
    label=r'NO$_2$',
    color='tab:green',
    linewidth=2,
    marker='^',
    markevery=marker_interval,
)

ax1.plot(
    time_in_hours,
    df['CO'],
    label=r'CO',
    #color='tab:green',
    linewidth=2,
    marker='^',
    markevery=marker_interval,
)

ax1.set_xlabel('Time of Day (HH:MM)', fontsize=12)
ax1.set_ylabel(
    'Concentration  [ppm]', fontsize=12
)

# 5. Fix X-Axis Ticks to proper hour intervals (e.g., every 1 or 2 hours)
hour_interval = 2  # Change to 1 for every hour, or 3 for every 3 hours
ax1.xaxis.set_major_locator(ticker.MultipleLocator(hour_interval))


# Function to format decimal hours into HH:MM (e.g., 7.5 -> 07:30)
def format_time(x, pos):
  hours = int(x) % 24
  minutes = int(round((x - int(x)) * 60))
  if minutes == 60:
    hours = (hours + 1) % 24
    minutes = 0
  return f'{hours:02d}:{minutes:02d}'


ax1.xaxis.set_major_formatter(ticker.FuncFormatter(format_time))

# 6. Secondary y-axis (twinx): Plot OH
ax2 = ax1.twinx()
ax2.plot(
    time_in_hours,
    df['OH'],
    label='OH',
    color='tab:red',
    linewidth=2,
    linestyle='--',
    marker='d',
    markevery=marker_interval,
)
ax2.set_ylabel('OH Concentration', fontsize=12, color='tab:red')
ax2.tick_params(axis='y', labelcolor='tab:red')

# 7. Force both y-axes to start at 0 and synchronize grid lines
ax1.set_ylim(bottom=0)
ax2.set_ylim(bottom=0)

num_bins = 5
locator1 = ticker.MaxNLocator(nbins=num_bins, prune='lower')
ax1.yaxis.set_major_locator(locator1)
ax2.yaxis.set_major_locator(
    ticker.LinearLocator(numticks=len(ax1.get_yticks()))
)

# Turn off secondary grid to prevent overlapping double-lines
ax2.grid(False)

# 8. Combine legends from both axes with a clean background frame
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(
    lines_1 + lines_2,
    labels_1 + labels_2,
    loc='upper left',
    frameon=True,
    facecolor='white',
    framealpha=0.9,
)

plt.title(
    r'Initial run',
    fontsize=14,
    pad=15,
)
plt.tight_layout()

# Save and show plot
plt.savefig('box_model_dual_axis_plot.png', dpi=300, bbox_inches='tight')
print("Plot saved successfully as 'box_model_dual_axis_plot.png'.")
plt.show()
```
